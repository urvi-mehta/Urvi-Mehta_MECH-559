```python
import numpy as np
import matplotlib.pyplot as plt

# Define the objective function
def f(x1, x2):
    return (x1 - x2)**2 + (x2 - 1)**2

# 1. Setup Grid data for contour plot
x1 = np.linspace(-0.5, 1.8, 400)
x2 = np.linspace(-0.5, 1.8, 400)
X1, X2 = np.meshgrid(x1, x2)
Z = f(X1, X2)

```

2. Define optimization paths from analytical results
Gradient method iterates: x0, x1, x2, x3


```python
gd_x1 = [0.0, 0.0, 0.5, 0.5]
gd_x2 = [0.0, 0.5, 0.5, 0.75]

# Newton's method iterates: x0, x1
newton_x1 = [0.0, 1.0]
newton_x2 = [0.0, 1.0]

# 3. Plotting
plt.figure(figsize=(8, 6.5))
contours = plt.contour(X1, X2, Z, levels=[0.05, 0.2, 0.5, 1.0, 2.0, 3.5, 5.0], colors='teal', linestyles='dashed', alpha=0.7)
plt.clabel(contours, inline=True, fontsize=9, fmt='%1.2f')

# Plot Gradient Descent Path
plt.plot(gd_x1, gd_x2, 'ro-', linewidth=2, markersize=6, label='Gradient Descent (Exact Line Search)')
for i, (x, y) in enumerate(zip(gd_x1, gd_x2)):
    plt.text(x - 0.05, y + 0.04, f'G{i}', color='red', weight='bold')

# Plot Newton's Method Path
plt.plot(newton_x1, newton_x2, 'bs--', linewidth=2, markersize=6, label="Newton's Method")
for i, (x, y) in enumerate(zip(newton_x1, newton_x2)):
    plt.text(x + 0.04, y - 0.06, f'N{i}', color='blue', weight='bold')

# Mark the Global Minimizer
plt.plot(1.0, 1.0, 'g*', markersize=14, label='Global Minimizer (1, 1)', zorder=5)

# Formatting
plt.xlim(-0.3, 1.7)
plt.ylim(-0.3, 1.7)
plt.xlabel('$x_1$', fontsize=12)
plt.ylabel('$x_2$', fontsize=12)
plt.title('Optimization Paths on Contour Plot of $f(x_1, x_2)$', fontsize=14, pad=15)
plt.legend(loc='upper left', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.5)

# Display the plot
plt.show()
```
